```python
import pandas as pd
import joblib

from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv("D:\Projects\lpg_flow_prediction\data\df_clean.csv")
```

```python
Distance = abs(df["Wthr"] - 2)

df["Closer"] = (Distance <= 2).astype(int)
```

```python
features = [
    "Stabilizer Feed T", "Stabilizer Feed Flow", "Stabilizer Top P", "Stabilizer Reflux Drum T", 
    "Stabilized Naphtha Flow", "Stabilizer Reflux Flow", "HGO CR Flow", "HGO CR to reboiler Flow","LPG Flow",
    "Stabilizer Top T",	"Stabilizer Top T.1",	"Stabilizer Bottom T",	"Stabiliser bottom level",
    "Stabilier bottom pressure", "Stab. 3rd Tray",	"Stab. 3rd Tray.1", "HGO CR Reboiler Inlet Temp( TI-1914)",	
    "Bottom Reboiler Naphtha Inlet Temp( TI-1907)",	"Bot. Naphtha Reboiler Outlet Temp(TI-1908 & TI-1909)",	
    "Bot. Naphtha Reboiler Outlet Temp(TI-1908 & TI-1909).1	", "Off Spec LPG flow (tag Faulty)", "Off Spec LPG from CRU inlet pressure",
    ]
```

```python
features = [f.strip() for f in features]
```

```python
X = df[features]

y = df["Closer"]
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    class_weight={0:1, 1:3.5},
    max_iter=1000,
    random_state=42
)

model.fit(X, y)
```

```python
joblib.dump(model, "D:\Projects\lpg_flow_prediction/backend/models/weathering_classifier.pkl")
joblib.dump(features, "D:\Projects\lpg_flow_prediction/backend/models/weathering_features.pkl")
```
